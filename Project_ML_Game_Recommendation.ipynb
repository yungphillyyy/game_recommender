```python
import pandas as pd

data = {
    'game': [
        'Minecraft',
        'Valorant',
        'Genshin Impact',
        'Call of Duty',
        'Stardew Valley',
        'PUBG',
        'Roblox'
    ],

    'genre': [
        'sandbox survival crafting',
        'fps shooter tactical',
        'openworld anime rpg',
        'fps shooter war',
        'farming relaxing simulation',
        'battle royale shooter',
        'sandbox multiplayer creativity'
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(df['genre'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(tfidf_matrix)
```

```python
def recommend_game(game_name, top_n=3):

    # Cari index game
    idx = df[df['game'] == game_name].index[0]

    # Ambil skor similarity
    sim_scores = list(enumerate(similarity[idx]))

    # Urutkan dari paling mirip
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Hilangkan game itu sendiri
    sim_scores = sim_scores[1:top_n+1]

    # Ambil nama game
    recommendations = []

    for i in sim_scores:
        recommendations.append(df.iloc[i[0]]['game'])

    return recommendations
```

```python
result = recommend_game("Valorant")

print("Recommended Games:")
for game in result:
    print("-", game)
```
